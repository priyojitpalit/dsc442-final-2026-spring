Write a function that:
1. Converts a word into a standard form (lowercase, no commas or periods)
2. Converts a full sentence into normalized senteence using the same rule.

```python
def normalize_word(w):
  return w.lower().strip(".,")

def normalize_sentence(s):
  return " ".join([normalize_word(w) for w in s.split()])
# normalize_word("Casino.") returns "casino"
# normalize_word("Car,") returns "car"
# normalize_word("HeLlO.") returns "hello"
# normalize_sentence("He bought a dozen of eggs. He also bought some chicken.")
# returns "he bought a dozen of eggs he also bought some chicken"
```

```python
normalize_word("Casino.")
normalize_sentence("He bought a dozen of eggs. He also bought some chicken.")
```

Write function that checks if an input is a valid zip code or not. Zip code are strings of length 5, consisting only of numbers.

```python
def isvalidzip(zip_code):
  return len(zip_code) == 5 and zip_code.isdigit()
```

```python
isvalidzip("12345")
```

```python
isvalidzip("a1234")
```

```python
isvalidzip("123456")
```
