```python
def demo (a, b, /, c, d=10, *args, e, f=20, **kwargs):
  """
  a, b are positional-only. Meaning, you need to enter just their values
  Writing a = something or b = something will throw an error.
  Anything before / is positional-only and anything after / is position or keyword based
  c, d, are both positional and keyword
  """
  print(f"a={a}")
  print(f"b={b}")
  print(f"c={c}")
  print(f"d={d}")
  print(f"args = {args}")
  for entry in args:
    print(entry)
  print(f"e = {e}")
  print(f"f = {f}")
  print(f"kwargs = {kwargs}")
```

```python
demo(2,3,4,e=5)
```

```python
demo(a = 2,3,4,e=5)
```

```python
demo(2,3, c = 4,e=5)
```

```python
demo(2,3, e=5, c=4)
```

```python
demo(2,3, 4, 5, 6, 7, 8, e = 9)
```

```python
demo(2,3, 4, 5, 6, 7, 8, e=9)
```

```python
demo(2,3, 4, 5, 6, 7, 8, e=9, g = 90, nonsense = 100, stuff = 909)
```

```python
demo(2,3, 4, 5, 6, 7, z = 8, e=9, g = 90, nonsense = 100, stuff = 909)
```

```python
def student (a, b, /, c, d, *, e, f ):
  print(f"a={a}")
  print(f"b={b}")
  print(f"c={c}")
  print(f"d={d}")
  print(f"e={e}")
  print(f"f={f}")
```

```python
student(1, 2, 3, 4, e = 5, f = 6)
```

```python
student(1, 2, 3, 4, 7, 8, e = 5, f = 6)
```

```python
def newstudent(a, b, /, c, d, *catch, e, f, **catch_all):
  print(f"a={a}")
  print(f"b={b}")
  print(f"c={c}")
  print(f"d={d}")
  print(f"catch={catch}")
  print(f"e={e}")
  print(f"f={f}")
  print(f"catch={catch_all}")
```

```python
newstudent(1,2,3,4,5,6,e=7,f=8,g=10)
```
