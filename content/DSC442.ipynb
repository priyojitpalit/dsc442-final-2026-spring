```python
def demo_best_model():
  models = [
      {"name": "A", "accuracy": 0.82},
      {"name": "B", "accuracy": 0.91},
      {"name": "C", "accuracy": 0.88}
  ]
  best = min(models, key = score)
  print(best)
#From the list of samples, find the one with the highest score
def score(sample):
  return sample["accuracy"]
demo_best_model()
```

```python
customer = [
    {"name" : "Alice", "total_spent": 1200, "visits":12},
    {"name" : "Bob", "total_spent": 800, "visits":4},
    {"name": "Carol", "total_spent": 1500, "visits":20}]

def best_customer(x):
  return x["total_spent"]/x["visits"]

# write a lambda function that does the same thing as best_customer

best_cust_lambda = lambda x:x["total_spent"]/x["visits"]

best = max(customer, key = best_customer)
print(best)
```

```python
#Sort customers by risk

customers = [
    {"name":"Alice", "debt":5000, "missed_payments": 1},
    {"name":"Carol", "debt":9000, "missed_payments": 0},
    {"name":"Bob", "debt":2000, "missed_payments": 3},

]
s = sum([cust['debt'] for cust in customers])

def risk(c):
    return c["missed_payments"]*2 + c["debt"] / s

best_customer = sorted(customers, key=risk, reverse = 1)
best_customer_names = [cust["name"] for cust in best_customer]
```
