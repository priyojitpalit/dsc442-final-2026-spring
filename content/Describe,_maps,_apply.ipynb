Convert in entry under points to point - mean(points)

```python
review_mean = reviews.points.mean()
def review_points_stnd(p):
  return p - review_mean
```

```python
stndpoints = reviews.points.map(review_points_stnd)
```

```python
def usd_cad(p):
  return f"CAD {p*1.37}"
priceincad = reviews.price.map(usd_cad)
priceincad
```

map() works on series, and apply() works on DataFrame

```python
def mean_points(row):
  row.points = row.points - review_mean
  return row

new_reviews = reviews.apply(mean_points, axis = 'columns')
new_reviews
```

```python
def premium_label(row):
  if row.points >= 95 and row.price > 50:
    return "Premium"
  else:
    return "Normal"

reviews['label'] = reviews.apply(premium_label, axis = 'columns')
```
