**apply()**

produces an entire dataframe

axis = 0 or axis = 'index' or nothing (column wise)

axis = 1 or axis = 'columns' (row wise)

```python
row = reviews.iloc[0]
row.points
```

```python
def remean_points(row):
  row.points = row.points - review_points_mean
  return row
```

```python
reviews.apply(remean_points, axis = 'columns')
```

```python
def remean_column(col):
  if col.name == 'points':
    return col - review_points_mean
  return col
reviews.apply(remean_column, axis = 'index')
```

```python
def remean_column(col):
  if col.name == 'points':
    return col - review_points_mean
  return col
reviews.apply(remean_column)
```

```python
def describe_column(col):
  if col.dtype != 'object':
    return col.describe()
reviews.apply(describe_column)
```

```python
def describe_column(col):
  return col.describe()
reviews.apply(describe_column)
```

```python
def normalize_numeric_only(col):
  if col.dtype != 'object':
    return col - col.mean()
  return col
reviews.apply(normalize_numeric_only)
```
